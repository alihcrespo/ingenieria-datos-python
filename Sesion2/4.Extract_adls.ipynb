```python
import pandas as pd
from azure.storage.filedatalake import DataLakeFileClient
from io import StringIO
```

```python
storage_account="sakrm001"  #nombre del datalake
sas_token ='' # cadena de conexion al latalake
container ="" #Contenedor del datalake
file_name ="" #archivo a buscar
```

```python
def obtener_cliente_adls():
    client = DataLakeFileClient(
        account_url=f"https://{storage_account}.dfs.core.windows.net", #url del datalake
        credential=sas_token
    )
    return client
```

```python
client = obtener_cliente_adls()
container_client = client.get_file_system_client(file_system=container)
file = container_client.get_file_client(file_name)
```

```python
download = file.download_file()
download_bytes = download.readall()
file_content = download_bytes.decode("utf-8")
file_content
```
